```python
import pandas as pd
import numpy as np
```

```python
cleaned_data = pd.read_csv('/Users/pradnyakolekar/Documents/MSc_Computing/Project/recommendation-system/output/netflix_final_cleaned.csv')
cleaned_data.head()
```

```python
# convert the date_added from text datetime format
cleaned_data["date_added"] = pd.to_datetime(cleaned_data["date_added"].astype(str).str.strip(), errors='coerce')
display(cleaned_data.head())
```

```python
# add columns for year and month added
cleaned_data["year_added"] = cleaned_data["date_added"].dt.year.astype('Int64')
cleaned_data["month_added"] = cleaned_data["date_added"].dt.month.astype('Int64')
display(cleaned_data.head())
```

```python
# ensure that release year as an integer values
cleaned_data["release_year"] = pd.to_numeric(cleaned_data["release_year"], errors='coerce')
```

```python
# split the duration for movies and shows into two columns
# For movies, the duration is in minutes, and for shows, the duration is in seasons

#extract the integer values from duration column
duration_values = pd.to_numeric(cleaned_data["duration"].astype(str).str.extract(r"(\d+)")[0], errors='coerce')

cleaned_data["duration_mins"] = np.where(cleaned_data["type"] == "Movie", duration_values, np.nan)
cleaned_data["seasons"] = np.where(cleaned_data["type"] == "TV Show", duration_values, np.nan)

# drop the original duration column
cleaned_data.drop(columns=["duration"])
```

```python
# renaming the column name for clarity
# rating -> maturity_rating
cleaned_data = cleaned_data.rename(columns={"rating": "maturity_rating"})
```

```python
# define the maturity rating in groups based on the age suitability
MATURITY_MAP = {
    # Kids
    "TV-Y": "Kids", "TV-G": "Kids", "G": "Kids",
    # Older Kids
    "TV-Y7": "Older Kids", "TV-Y7-FV": "Older Kids", "PG": "Older Kids",
    # Teens
    "PG-13": "Teens", "TV-PG": "Teens", "TV-14": "Teens",
    # Adults
    "R": "Adults", "NC-17": "Adults", "TV-MA": "Adults",
    # Unrated
    "NR": "Unrated", "UR": "Unrated",
    # Unknown
    "Unknown": "Unknown"
}

# create a maturity_group column based on the maturity rating
cleaned_data["maturity_group"] = cleaned_data["maturity_rating"].astype(str).map(MATURITY_MAP).fillna("Unknown")

# check the changes
display(cleaned_data.head())
```

```python
# split the multi-valued data into lists for genres, directors, cast, and countries
# keeping empty list for missing values and "Unknown" values

def to_list(val, max_items=None):
    if pd.isna(val) or str(val).strip() in ["Unknown", ""]:
        return []
    items = [item.strip() for item in str(val).split(",") if item.strip()]
    return items[:max_items] if max_items else items

# Create the new list columns
cleaned_data["genres"] = cleaned_data["listed_in"].apply(to_list)
cleaned_data["directors"] = cleaned_data["director"].apply(to_list)
cleaned_data["cast_list"] = cleaned_data["cast"].apply(lambda x: to_list(x, max_items=3))
cleaned_data["countries"] = cleaned_data["country"].apply(to_list)

# Verify the output
cleaned_data[["title", "genres", "directors", "cast_list", "countries"]].head()
```

```python
#Primary Country
# get first country from the list, or 'Unknown' if the list is empty
raw_primary = cleaned_data["countries"].apply(lambda lst: lst[0] if len(lst) > 0 else "Unknown")

# countries with less than 20 titles (exclude 'Unknown')
country_counts = raw_primary[raw_primary != "Unknown"].value_counts()
rare_countries = set(country_counts[country_counts < 20].index)

# Group rare countries into 'Other', keeping 'Unknown' as 'Unknown'
cleaned_data["primary_country"] = raw_primary.apply(lambda c: "Other" if c in rare_countries else c)

# Check the distribution and sample rows
print(cleaned_data["primary_country"].value_counts())
cleaned_data[["title", "country", "countries", "primary_country"]].head(10)
```

```python
# expected columns
expected_cols = [
    "show_id", "type", "title", "director", "cast", "country", "listed_in", "description", "release_year",
    "date_added", "year_added", "month_added", "duration_mins", "seasons",
    "maturity_rating", "maturity_group", "genres", "directors", "cast_list", "countries", "primary_country"
]

# Check for any missing or extra columns
missing = set(expected_cols) - set(cleaned_data.columns)
extra = set(cleaned_data.columns) - set(expected_cols)
print("Missing columns:", missing)
print("Extra columns:", extra)

# Save Parquet preserves lists and dates cleanly
cleaned_data.to_parquet("parsed_netflix_data.parquet", index=False)
print("Saved parsed_netflix_data.parquet successfully. Shape:", cleaned_data.shape)
```

```python
# save CSV for compatibility, but lists will be stored as strings
cleaned_data.to_csv("parsed_netflix_data.csv", index=False)
```
